# VLAB_02 — Two-step API Showcase

Demonstrates the `generate_network` / `simulate_from_network` split:

1. Generate **9 networks** once (3 inlet seeds × 3 network generation seeds).
2. For each network, run a number of **MODFLOW realizations** varying only aquifer properties (hk, ss, sy) while holding epikarst (recharge) fixed.
3. Show a **3×3 grid of head-uncertainty maps** (std of heads at the 3rd transient stress period).

```python
import vlab
import numpy as np
import matplotlib.pyplot as plt
```

## Configuration

```python
INLET_SEEDS = [100, 456, 789]       # 3 columns — different inlet placements
SKS_SEEDS   = [12345, 5000, 25000]  # 3 rows    — different network realisations
N_AQUIFER   = 10                    # LHS realisations of (hk, ss, sy) per network
N_SP        = 5                     # transient stress periods
T_IDX       = 3                     # heads index for the 3rd transient period
NROW        = 20                    # number of rows in the model domain
NCOL        = 40                    # number of columns in the model domain
```

## Helper

```python
def make_system(inlet_seed, sks_seed):
    """Build a System with a fixed network topology and stochastic aquifer params."""
    return vlab.System(
        n_rows=NROW,
        n_cols=NCOL,
        n_stress_periods=N_SP,
        network_structure=vlab.NetworkStructure(
            stochastic=False,
            values={"sks_seed": sks_seed},
            inlets_seed=inlet_seed,
        ),
        recharge_params=vlab.RechargeParams(stochastic=False),
        network_properties=vlab.NetworkProperties(stochastic=False),
        aquifer_params=vlab.AquiferParams(stochastic=True),
    )
```

## Step 1 — Generate networks

Each of the 9 networks is generated once and stored.  
The MODFLOW loop (Step 2) reuses them without re-running pyKasso.

```python
networks = {}  # (row, col) → vlab.Network

for row, sks in enumerate(SKS_SEEDS):
    for col, ins in enumerate(INLET_SEEDS):
        s = make_system(ins, sks)
        networks[(row, col)] = s.generate_network(random_state=None, cleanup=True)
        print(f"  sks={sks:>6}  inlet_seed={ins}  → {len(networks[(row, col)].idxs_inlet)} inlets")
```

### Stochastic Networks — 3×3 overview

```python
fig, axes = plt.subplots(3, 3, figsize=(12, 6))

for row, sks in enumerate(SKS_SEEDS):
    for col, ins in enumerate(INLET_SEEDS):
        ax = axes[row, col]
        net = networks[(row, col)]

        nplot = net.validator.network.copy().astype(float)
        nplot[nplot == 0] = np.nan
        ax.matshow(nplot, cmap="cividis", vmin=2, vmax=11)

        inlets = np.array(net.idxs_inlet)
        ax.scatter(
            inlets[:, 0] - 1, inlets[:, 1] - 1,
            c="r", marker="v", edgecolors="k", s=25, zorder=5,
        )
        ax.scatter(
            [net.idxs_spring[0]], [net.idxs_spring[1]],
            c="cyan", marker=">", edgecolors="k", s=25, zorder=5,
        )

        ax.set_xticks([])
        ax.set_yticks([])
        if row == 0:
            ax.set_title(f"inlet_seed = {ins}", fontsize=9)
        if col == 0:
            ax.set_ylabel(f"sks = {sks}", fontsize=9)

plt.suptitle("Stochastic karst networks", fontsize=11)
plt.tight_layout()
plt.show()
```

## Step 2 — MODFLOW ensemble

For each of the 9 networks: 10 LHS realisations of `(hk, ss, sy)` with fixed epikarst recharge.  
Total: **90 MODFLOW runs**.

```python
heads_all = {}  # (row, col) → np.ndarray shape (n_valid, n_rows, n_cols)
discharge_all = {}  # (row, col) → np.ndarray shape (n_valid, n_rows, n_cols)

for row, sks in enumerate(SKS_SEEDS):
    for col, ins in enumerate(INLET_SEEDS):
        s = make_system(ins, sks)
        s.generate_samples(n_samples=N_AQUIFER, seed=1)
        net = networks[(row, col)]

        heads_list = []
        discharge_list = []
        for rs in range(N_AQUIFER):
            success, discharge, heads = s.simulate_from_network(
                network=net, random_state=rs, cleanup=True
            )
            if success:
                heads_list.append(heads[T_IDX, 0, :, :])
                discharge_list.append(discharge)

        heads_all[(row, col)] = np.array(heads_list)
        discharge_all[(row, col)] = np.array(discharge_list)
        print(
            f"  sks={sks:>6}  inlet_seed={ins}  "
            f"→ {len(heads_list)}/{N_AQUIFER} successful runs"
        )
```

## Step 3 — Head uncertainty

Standard deviation of heads at the 3rd transient stress period across the 10 aquifer realisations.

```python
std_maps = {k: np.std(v, axis=0) for k, v in heads_all.items()}
vmax = max(m.max() for m in std_maps.values())

fig, axes = plt.subplots(3, 3, figsize=(12, 6))

for row in range(3):
    for col in range(3):
        ax = axes[row, col]
        im = ax.matshow(std_maps[(row, col)], cmap="magma", vmin=0, vmax=vmax)
        ax.set_xticks([])
        ax.set_yticks([])
        if row == 0:
            ax.set_title(f"inlet_seed = {INLET_SEEDS[col]}", fontsize=9)
        if col == 0:
            ax.set_ylabel(f"sks = {SKS_SEEDS[row]}", fontsize=9)

fig.colorbar(im, ax=axes, shrink=0.6, label="Std of heads [m]")
plt.suptitle(
    f"Head uncertainty (std) at t={T_IDX}  |  {N_AQUIFER} aquifer realisations per network",
    fontsize=11,
)
# plt.tight_layout()
plt.show()
```

## Step 4 — Spring discharge uncertainty

Standard deviation of spring discharge time series / different realizations.

```python
ts_vals = {k: v for k, v in discharge_all.items()}
vmax = max(m.max() for m in ts_vals.values())
vmin = min(m.min() for m in ts_vals.values())

fig, axes = plt.subplots(3, 3, figsize=(12, 6), sharex=True, sharey=True)

for row in range(3):
    for col in range(3):
        ax = axes[row, col]
        im = ax.plot(ts_vals[(row, col)], c="k", lw=1.)
        ax.set_xlim(0, N_SP)
        ax.set_ylim(vmin, vmax)
        ax.set_yscale("log")
        if row == 0:
            ax.set_title(f"inlet_seed = {INLET_SEEDS[col]}", fontsize=9)
        if col == 0:
            ax.set_ylabel(f"sks = {SKS_SEEDS[row]}", fontsize=9)

plt.suptitle(
    f"Discharge uncertainty |  {N_AQUIFER} aquifer realisations per network",
    fontsize=11,
)
# plt.tight_layout()
plt.show()
```
